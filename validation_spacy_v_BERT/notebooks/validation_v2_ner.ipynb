```python
# =============================================================================
# CELL 7: DISPLAY RESULTS
# =============================================================================

import pandas as pd

# Re-establish TEST_MODE context for standalone execution
TEST_MODE = os.environ.get('TEST_MODE', 'False').lower() == 'true'
SESSION_ID = os.environ.get('SESSION_ID', '')
output_suffix = f"_{SESSION_ID}" if SESSION_ID else ("_test" if TEST_MODE else "")

print("📊 NER Results Summary\n")
print("="*70)

# Load results
results_file = Path(f"results/validation/ner/v2_ner_results{output_suffix}.csv")

if results_file.exists():
    df_results = pd.read_csv(results_file)
    
    print(f"\n📄 Results file: {results_file}")
    print(f"📦 File size: {results_file.stat().st_size / 1024:.1f} KB")
    
    print(f"\n📊 Entity Extraction Statistics:")
    print(f"   Total entities extracted: {len(df_results)}")
    print(f"   Papers with entities: {df_results['ID'].nunique() if 'ID' in df_results.columns else 'N/A'}")
    
    # Show entity type distribution
    if 'label' in df_results.columns:
        print(f"\n📈 Entity Types:")
        for label, count in df_results['label'].value_counts().items():
            print(f"   {label}: {count} ({100*count/len(df_results):.1f}%)")
    
    # Show sample entities
    print(f"\n📋 Sample Entities (first 10):")
    display_cols = ['ID', 'mention', 'label']
    if 'start_char' in df_results.columns:
        display_cols.append('start_char')
    if 'end_char' in df_results.columns:
        display_cols.append('end_char')
    
    available_cols = [col for col in display_cols if col in df_results.columns]
    print(df_results[available_cols].head(10))
    
    # Show unique entities (memory-safe)
    if 'mention' in df_results.columns:
        unique_count = df_results['mention'].nunique()
        print(f"\n🔍 Unique Entities:")
        print(f"   Total unique mentions: {unique_count}")
        
        # Memory-safe display of top entities
        MAX_DISPLAY = 10
        if unique_count > 0:
            print(f"\n   Top {MAX_DISPLAY} most frequent:")
            # Use nlargest() which is more memory efficient for large datasets
            top_entities = df_results['mention'].value_counts().nlargest(MAX_DISPLAY)
            for mention, count in top_entities.items():
                print(f"      {mention}: {count} occurrences")
            
            if unique_count > MAX_DISPLAY:
                print(f"   ... and {unique_count - MAX_DISPLAY} more unique entities")
        
        # Warning if entity count is suspiciously low
        if len(df_results) < 10:
            print("\n⚠️ WARNING: Very few entities extracted - model may have issues")
            print("   Consider checking model loading and input format")
    
    print(f"\n✅ Results successfully saved to Drive")
    print(f"   Location: {results_file}")
    
else:
    print(f"⚠️ Results file not found: {results_file}")
    print("NER may have failed - check output above")

print("\n" + "="*70)
```
